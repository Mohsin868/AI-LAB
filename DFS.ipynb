```python
tree = {
    "A":["B", "C"],
    "B":["D", "E"],
    "C":["F"],
    "D":[],
    "E":[],
    "F":["G"],
    "G":[]
}

def dfs_stack(start, goal):
    visited = []
    stack = [start]   

    while stack:
        current = stack.pop() 

        if current not in visited:
            visited.append(current)

            if current == goal:
                print(current)  
                break

            for child in reversed(tree[current]):
                stack.append(child)

    return visited

print(dfs_stack("A", "F"))
```

🌳 DFS Traversals in Binary Trees

DFS (Depth First Search) means exploring a tree deeply before moving to the next branch.
There are 3 main types of DFS traversals:

✅ 1. Inorder (Left → Root → Right)

Visit left subtree

Visit root (current node)

Visit right subtree

📌 Example order: D → B → E → A → C
📊 Use: In Binary Search Trees, gives elements in sorted order.

✅ 2. Preorder (Root → Left → Right)

Visit root (current node)

Visit left subtree

Visit right subtree

📌 Example order: A → B → D → E → C
📊 Use: Used to copy or serialize a tree.

✅ 3. Postorder (Left → Right → Root)

Visit left subtree

Visit right subtree

Visit root (current node)

📌 Example order: D → E → B → C → A
📊 Use: Used to delete a tree or evaluate expression trees.

🧠 Quick Comparison
Traversal	Order	Root visited	Common Use
Inorder	Left → Root → Right	2nd	Sorted order in BST
Preorder	Root → Left → Right	1st	Copying / serializing tree
Postorder	Left → Right → Root	3rd	Deleting / evaluating expressions
