```python
tree = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": ["G"],
    "G": []
}

def bfs(start, goal):
    visited = []           
    container = [start]    

    while container:      
        current = container[0]      
        container = container[1:]   

        if current not in visited:
            visited.append(current)

            if current == goal:
                print(current)
                return visited

            for neighbor in tree[current]:
                container.append(neighbor)

    return visited

bfs("A", "F")
```

```python
tree = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": ["G"],
    "G": []
}

visited = []

def bfs_queue(start, goal):
    queue = [start]   

    while queue:
        current = queue.pop(0)   

        if current not in visited:
            visited.append(current)

            if current == goal:
                print(current)
                break

            for neighbor in tree[current]:
                queue.append(neighbor)

    return visited

print(bfs_queue("A", "F"))
```
