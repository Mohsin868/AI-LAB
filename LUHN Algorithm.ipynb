Code for LUHN Algorithm

16 digits card number

checks very quickly if card is valid or not

5 8 9 3 8 0 4 1 1 5 4 5 7 2 8 9

Step 1: Remove the last digit

5 8 9 3 8 0 4 1 1 5 4 5 7 2 8 x

Step 2: Reverse

8 2 7 5 4 5 1 1 4 0 8 3 9 8 5

Step 3: Double digits that are at even index

16 2 14 5 8 5 2 1 8 0 16 3 18 8 10

Step 4 : Minus any number larger than 9

7 2 5 5 8 5 2 1 8 0 7 3 9 8 1 x

Step 5: Add all numbers

sum + x

71+9 = 80

Step 6:
n % 10 == 0:

then True

```python
# Checking Card Number Validity
def check_card_validity(card_number):
    # Step 1: Remove the last digit (check digit)
    check_digit = int(card_number[-1])
    print("\n")
    print("Step 1:")
    print("Remove Last Digit:", check_digit)
    print("Remaining Digits:", card_number[:-1])

    # Step 2: Reverse the digits
    card_number = list(map(int, card_number[:-1]))
    card_number.reverse()
    print("\n")
    print("Step 2:")
    print("Reversed Digits:", card_number)

    # Step 3: Double digits at even index
    for i in range(0, 15 , 2):
        card_number[i] *= 2
    print("\n")
    print("Step 3:")
    print("After doubling:", card_number)


    # Step 4 Minus any number larger than 9
    for i in range(0, 15 , 2):
        if card_number [i] > 9:
            card_number[i] -= 9
    print("\n")
    print("Step 4:")
    print("After subtracting 9 from numbers > 9:", card_number)

    # Step 5: Add all numbers + check digit
    total = sum(card_number) + check_digit
    print("\n") 
    print("Step 5:")
    print("Total Sum including last digit:", total)   

    # Step 6: Check divisibility by 10
    print("\n")
    print("Step 6:")
    if total % 10 == 0:
        print("Total divisible by 10")
        return True
    else:
        print("Total not divisible by 10")
        return False



# Example usage
card = "4539148803436467"

print("Card Number:", card)
# Call the function
if check_card_validity(card):
    print("Valid card number ")
else:
    print("Invalid card number ")
```
