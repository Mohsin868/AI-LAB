Write a python program to sort the sentence in alphabetical order?

```python
def bubble_sort(chars):
    n = len(chars)
    for i in range(n):
        for j in range(0, n - i - 1):
            if ord(chars[j]) > ord(chars[j + 1]):  
                chars[j], chars[j + 1] = chars[j + 1], chars[j]
    return chars

chars = [s for s in "I am Rasikh" ]

# Sort
sorted_chars = bubble_sort(chars)

# Print result
print("".join(sorted_chars))
```

```python
def bubble_sort(chars):
    n = len(chars)
    for i in range(n):
        for j in range(0, n - i - 1):
            if ord(chars[j]) > ord(chars[j + 1]):  # compare ASCII values
                chars[j], chars[j + 1] = chars[j + 1], chars[j]
    return chars

def Sorted(sentence):
    chars = [c for c in sentence ]
    
    # Sort using bubble_sort
    sorted_chars = bubble_sort(chars)
    
    # Show ASCII values for each char
    for c in sorted_chars:
        print(f"{c}: {ord(c)}")
    
    return ''.join(sorted_chars)


# Example
input_sentence = "I am Rasikh"
sorted_chars = Sorted(input_sentence)
print("\nSorted Characters:", sorted_chars)
```

```python
# Sort by Character ASCII Value

def sort_by_ascii(s):
    return ''.join(sorted(s, key=lambda char: ord(char)))

sorted_str = sort_by_ascii(str)
print( sorted_str)
```

```python
# Sort without built in functions

def sort(arr):
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] > arr[j]:
                arr[i], arr[j] = arr[j], arr[i]
    return arr

print(sort([5, 3, 8, 6, 7, 2]))  
```
