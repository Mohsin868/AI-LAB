Punctuation / Special Characters

Remove punctuations from a string

```python
def remove_punctuations(text):
    punctuations = '''!()-[]{};:'",<>./?@#$%^&*_~'''
    result = ""
    for char in text:
        if char not in punctuations:
            result += char
    return result

# Example usage
sentence = "Hello! How are you doing today? Let's learn Python."
print("Original:", sentence)
print("Without Punctuations:", remove_punctuations(sentence))
```

```python
class TextCleaner:
    def __init__(self, text):
        self.text = text

    def remove_punctuations(self):
        punctuations = '''!()-[]{};:'",<>./?@#$%^&*_~'''
        cleaned = ""
        for char in self.text:
            if char not in punctuations:
                cleaned += char
        return cleaned

# Example usage
sentence = "Hello! How are you doing today? Let's learn Python."
cleaner = TextCleaner(sentence)
print("Original:", sentence)
print("Without Punctuations:", cleaner.remove_punctuations())
```
