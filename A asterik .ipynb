## AI LAB 7 – A* (A Star) Algorithm

#### The A* algorithm finds the shortest path from a start node to a goal node
#### using both the actual cost (g) and estimated cost (h).

```python
import heapq  

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [],
    'E': [('F', 1)],
    'F': []
}

# Heuristic values (estimated cost to goal)
heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 1,
    'F': 0
}

def a_star(start, goal):
    # Priority queue (f, node, path)
    pq = []
    heapq.heappush(pq, (0, start, [start]))

    # g-costs (distance from start)
    g_cost = {start: 0}

    while pq:
        f, current, path = heapq.heappop(pq)

        if current == goal:
            print("Visited path:", path)
            return g_cost[current]

        for neighbor, cost in graph[current]:
            new_g = g_cost[current] + cost
            f_new = new_g + heuristic[neighbor]

            # If better path found
            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                heapq.heappush(pq, (f_new, neighbor, path + [neighbor]))

    return None

# Run A*
cost = a_star('A', 'F')
print("Total cost from A to F:", cost)
```
