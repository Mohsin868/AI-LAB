• Model-Based Reflex Agent

• This agent not only checks the current temperature but also remembers the previous 
action to avoid turning the heater on or off unnecessarily.

```python
class ModelBasedReflexAgent:
    def __init__(self, fixed_temp):
        self.fixed_temp = fixed_temp
        self.current_temp = None
        self.memory = {}   #  To store previously done actions
    
    def sensor(self, temp):
        self.current_temp = temp
    
    def performance(self):
        # If already in memory, reuse the action from the memory 
        if self.current_temp in self.memory:
            print(f"Memory found for {self.current_temp}°C  Using stored action")
            return self.memory[self.current_temp]
        
        # Otherwise perform new action
        print(f"First time at {self.current_temp}°C ")
        #  Doing full process and storing in memory"

        if self.current_temp > self.fixed_temp:
            action = "Turn on the AC"
        else:
            action = "Turn off the AC"
        
        # Save action in memory
        self.memory[self.current_temp] = action
        return action
    
    def actuator(self):
        action = self.performance()
        print(self.current_temp, " Action =>:", action)


# Create agent with desired fixed temperature
agent = ModelBasedReflexAgent(26)

# Example single check
agent.sensor(24)
agent.actuator()
print("Fixed Temp =>:", agent.fixed_temp, "| Current Temp =>:", agent.current_temp)

# Rooms dictionary
rooms = {
    "Living Room": 30, 
    "Drawing Room": 24,
    "Bed Room": 22,
    "Kitchen": 34,
    "Store Room": 20,
    "Living Room Again": 30,  # to test memory
    "Bed Room Again": 22      # to test memory
}

# Testing each room
for room, temp in rooms.items():
    print(room, end=":\t")
    agent.sensor(temp)
    agent.actuator()
```
