TASK 2 MOVIES


```python
movies = [
    ("Eternal Sunshine of the Spotless Mind", 20000000),
    ("Memento", 9000000),
    ("Requiem for a Dream", 4500000),
    ("Pirates of the Caribbean: On Stranger Tides", 379000000),
    ("Avengers: Age of Ultron", 365000000),
    ("Avengers: Endgame", 356000000),
    ("Incredibles 2", 200000000)
]
```

```python
# Calculate total and then average budget of the movies
total_budget = 0
for movie in movies:
    total_budget += movie[1]  

average_budget = total_budget / len(movies)

print("\n")
print(f"Total Budget: ${total_budget}")
print("\n")
print(f"Average Budget: ${average_budget}")
```

```python
# Print movies above average
above_average = []
for movie in movies:
    name = movie[0]
    budget = movie[1]
    if budget > average_budget:
        difference = budget - average_budget
        print(name, "had a budget $", round(difference), "higher than average.")
        above_average.append(name)

# Print how many movies exceeded the average budget
print("Number of movies above average:", len(above_average))
```

TASK 1 FIZZ BUZZ

```python
import random

def fizzbuzz_game(max_range=50):
    score = 0
    prev_number = 0

    print("\n")
    print("Welcome to  FizzBuzz Game!")
    print("Rules:")
    print(" Type 'Fizz' if divisible by 3")
    print(" Type 'Buzz' if divisible by 5")
    print(" Type 'Fizz Buzz' if divisible by both")
    print(" Press Enter if not divisible by 3 or 5")
    print(" Game ends if your answer is wrong")
    print("\n")

    while True:
        number = random.randint(1, max_range)
        print(f"Player 1 says: {number}")


        if number % 3 != 0 and number % 5 != 0:
            if prev_number != 0:
                number = prev_number + number
                print(f"Not divisible add with previous new number = {number}")
            else:
                print("Not divisible no previous number to add")


        if number % 3 == 0 and number % 5 == 0:
            correct_answer = "Fizz Buzz"
        elif number % 3 == 0:
            correct_answer = "Fizz"
        elif number % 5 == 0:
            correct_answer = "Buzz"
        else:
            correct_answer = ""

        # Player 2 input
        player_answer = input("Your answer: ").strip().capitalize()

        
        if player_answer == correct_answer:
            print(" Correct!")
            score += 1
        else:
            print(f" Wrong Answer! Expected: '{correct_answer}' but you said: '{player_answer}'")
            break

        prev_number = number  

    print(f"Final Score: {score}")


fizzbuzz_game(50)
```
